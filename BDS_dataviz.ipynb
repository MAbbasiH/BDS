### Why Visualize? The Power of Exploratory Data Analysis (EDA)

Before we can build sophisticated machine learning models or derive definitive business conclusions, we must first understand our data. This crucial, initial phase of any data science project is called **Exploratory Data Analysis (EDA)**. EDA is an approach to analyzing datasets to summarize their main characteristics, often with visual methods. It's about getting to know your data intimately, much like a detective investigating a crime scene. You're looking for clues, patterns, strange occurrences (anomalies), and relationships between different variables.

Think about it this way: if you were given a 1,000-page book and asked to summarize its plot, you wouldn't just start reading from page one and hope to remember everything. You might first look at the cover, read the back-of-the-book summary, scan the table of contents, and perhaps flip through to look at illustrations. This is what EDA is for data.

Summary statistics like mean, median, and standard deviation are like the back-of-the-book summary—they are incredibly useful but can also be misleading. The famous **Anscombe's Quartet** is a classic example where four datasets have nearly identical summary statistics, yet look wildly different when plotted. This is where data visualization becomes not just a tool, but an essential skill.

Let's take a practical look at **Anscombes Quartet**, as the dataset happen to be built in to our tool of choice.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Seaborn has the Anscombe dataset built-in, which makes it very easy to load.
anscombe = sns.load_dataset("anscombe")

# --- 1. PROVE THE STATISTICS ARE MISLEADING ---
# First, let's prove that the summary statistics for each of the four datasets are nearly identical.
# We'll group the data by the 'dataset' column (I, II, III, IV) and calculate key stats.

# Select only the columns we need for aggregation
agg_cols = ['x', 'y']

# Use .agg() to calculate multiple statistics at once
summary_stats = anscombe.groupby('dataset')[agg_cols].agg(
    ['mean', 'std', 'var']
).round(2) # Rounding for cleaner display

# Calculate the correlation for each dataset separately
correlations = anscombe.groupby('dataset')[['x', 'y']].corr().unstack().iloc[:, 1].rename('correlation').round(2)

# Combine the statistics into a single summary table
summary_table = pd.concat([summary_stats, correlations], axis=1)

print("--- Misleading Summary Statistics ---")
print("Notice how these values are nearly identical for all four datasets:")
print(summary_table)
print("\n" + "="*40 + "\n")


# --- 2. REVEAL THE TRUTH WITH VISUALIZATION ---
# Now, we'll use Seaborn's `lmplot` to create the four plots.
# `lmplot` is perfect for this because it can create a scatter plot,
# add a linear regression line, and create subplots (facets) for different
# categories all in one command.

print("--- Revealing the Truth with Plots ---")
print("Now, let's visualize the datasets to see their true, different stories.")
print("-"*40)

# Set the overall theme for the plots
sns.set_theme(style="ticks")

# Create the faceted scatter plots
# - `x` and `y` define the axes.
# - `col="dataset"` tells Seaborn to create a separate plot for each unique value in the 'dataset' column.
# - `hue="dataset"` gives each plot a different color.
# - `col_wrap=2` arranges the four plots into a 2x2 grid.
# - `ci=None` turns off the confidence interval shading around the regression line.
# - `height=4` sets the size of each plot.
g = sns.lmplot(
    data=anscombe,
    x="x",
    y="y",
    col="dataset",
    hue="dataset",
    col_wrap=2,
    ci=None,
    palette="muted",
    height=4,
    scatter_kws={"s": 50, "alpha": 1} # Makes the points slightly larger
)

# Add a descriptive title for the entire figure
g.fig.suptitle("Anscombe's Quartet: The Importance of Visualization", y=1.03)

# Show the plot
plt.show()
```

A well-crafted plot can reveal:

* **Distributions:** How are the values of a single variable spread out? Is there a common value? Are there multiple peaks?
* **Relationships:** How does one variable change as another changes? Is there a positive or negative correlation?
* **Outliers:** Are there any data points that are unusually different from the rest? These could be errors or critically important findings.
* **Patterns and Structures:** Are there clusters of data points? Are there trends over time?

In this workshop, we will learn how to use visualization as our primary tool for conducting effective EDA. You will learn to move from raw data to insightful graphics that tell a compelling story.

### Essential Setup: Importing Libraries and Setting the Theme

Before we can create our first plot, we need to set up our programming environment. This involves importing the necessary libraries and configuring our plotting style. This is a standard ritual that you will perform at the beginning of almost every data analysis notebook.

After importing the libraries, we will call `sns.set_theme()`. This is a simple but powerful function that applies Seaborn's beautiful default styling to all subsequent plots we create, even those made with pure Matplotlib. ### Visualizing a Single Numerical Variable: The Histogram

The most common and intuitive way to visualize the distribution of a single numerical variable is the **histogram**. A histogram works by grouping the data into a series of intervals, or "bins," of equal width. It then plots the number of observations (the frequency) that fall into each bin. This allows us to see the shape of the data's distribution. Seaborn automatically chooses a reasonable number of bins based on the data's characteristics (often using an algorithm like the Freedman-Diaconis rule), but you can explicitly control this with the `bins` parameter, as we'll see shortly.

Let's create a histogram for the `flipper_length_mm` column to understand how penguin flipper lengths are distributed across our dataset. We'll use Seaborn's `histplot` function, which is specifically designed for this purpose.

```python
# Create a histogram of the flipper_length_mm column.
# The `data` argument specifies the DataFrame to use.
# The `x` argument specifies the column to plot on the x-axis.
sns.histplot(data=penguins, x="flipper_length_mm")

# Add a title for clarity using matplotlib's function
plt.title("Distribution of Penguin Flipper Lengths")

# Display the plot
plt.show()
```

#### Interpretation

This simple plot is already packed with information:

* **X-axis (flipper_length_mm):** This represents the values of the variable we are analyzing—the length of the penguins' flippers in millimeters. We can see the range is roughly from 170 mm to 230 mm.
* **Y-axis (Count):** This shows the frequency, or the number of penguins that have a flipper length falling within the range of each bar (bin).
* **Shape of the Distribution:** This is the most important insight. The distribution is clearly not a simple bell curve (a "normal" distribution). We see two distinct peaks, which we call a **bimodal distribution**. This strongly suggests that there might be two different subgroups of penguins within our dataset that have systematically different flipper lengths. A very likely explanation is that the different species of penguins have different average flipper lengths.

#### The Impact of Bin Selection
While histograms are excellent, their appearance can be sensitive to the number of bins. A different choice of bins can produce a different-looking plot, potentially leading to different interpretations. Let's demonstrate this by plotting the same data with a very small and a very large number of bins.

```python
# Create a figure with two subplots side-by-side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Too few bins (bins=10)
sns.histplot(data=penguins, x="flipper_length_mm", bins=10, ax=axes[0])
axes[0].set_title("Flipper Lengths with Too Few Bins (10)")

# Plot 2: Too many bins (bins=100)
sns.histplot(data=penguins, x="flipper_length_mm", bins=100, ax=axes[1])
axes[1].set_title("Flipper Lengths with Too Many Bins (100)")

plt.suptitle("The Impact of Bin Selection on Histogram Shape")
plt.show()
```

As you can see, with too few bins, we lose detail and the bimodal nature is less clear. With too many bins, the plot becomes noisy and the overall shape is obscured by small fluctuations. This sensitivity is a key reason to also use a complementary visualization like a KDE plot.

### Smoothing it Out: The Kernel Density Estimate (KDE)

An alternative and complementary way to view the distribution is the **Kernel Density Estimate (KDE)** plot. You can think of a KDE plot as a smoothed version of the histogram. It represents the data using a continuous probability density curve, which can make it easier to see the underlying shape of the distribution, free from the "noise" of individual bins.

Just as a histogram's shape depends on the `bins` parameter, a KDE's shape depends on a smoothing parameter called **bandwidth** (controlled by `bw_adjust` in Seaborn). A larger bandwidth creates a smoother curve (like having fewer bins), while a smaller bandwidth creates a more "jagged" curve that follows the data more closely (like having many bins). Seaborn's default bandwidth selection (often based on rules like Scott's or Silverman's) is usually a good starting point, but understanding its role is key to interpreting the smoothness of the curve.

Seaborn makes it incredibly easy to overlay a KDE plot on top of our histogram. We simply add the argument `kde=True`.

```python
# Create the same histogram, but this time add a KDE curve.
sns.histplot(data=penguins, x="flipper_length_mm", kde=True)

# Add a title
plt.title("Distribution of Flipper Lengths with KDE")

# Display the plot
plt.show()
```

#### Interpretation

The resulting plot shows our original histogram with a smooth line drawn over it. This KDE line confirms our earlier observation from the histogram: there are two clear peaks in the data, reinforcing the bimodal nature of the flipper length distribution. The smooth curve makes the separation between the two groups even more visually apparent.

### A More Versatile Tool: `sns.displot`

So far, we've used `histplot`. Seaborn has another, more powerful function called `displot` (distribution plot) that is highly recommended. While `histplot` is an "axes-level" function (it draws onto a single set of axes), `displot` is a "figure-level" function. This means it creates and manages its own figure.

This distinction is crucial. Because `displot` is a figure-level function, it consistently returns a `FacetGrid` object. This object provides a unified API for customization and, most importantly, allows for the creation of multiple subplots based on categorical variables (a technique called "faceting") using the `col` or `row` parameters. We will see the power of this later. For now, just know that `displot` is the more robust and future-proof choice.

For a simple histogram, the syntax is very similar. We just specify `kind="hist"`.

```python
# Use the more flexible `displot` function to create the same plot.
# We specify the kind of plot we want with the `kind` parameter.
sns.displot(data=penguins, x="flipper_length_mm", kind="hist", kde=True)

# Add a title using a slightly different method for figure-level plots
plt.suptitle("Distribution of Flipper Lengths using displot", y=1.02) # y as a parameter adjusts title position
plt.xlabel("This is my x-label")

# Display the plot
plt.show()
```

As you can see, the output is nearly identical. For now, the key takeaway is that `displot` is a more versatile function that you should prefer for plotting distributions, as it will allow you to do more advanced things later on.

## 3. Exploring Relationships with Seaborn: From One to Many Variables

In the previous section, we mastered **univariate analysis**, focusing on the distribution of a single variable at a time. This was a crucial first step, and it already yielded a powerful insight: the distributions for flipper length and body mass were bimodal, suggesting the presence of distinct subgroups in our data.

Now, we move to the next logical step in Exploratory Data Analysis: **bivariate analysis**. Here, our goal is to explore the relationship *between two variables*. This is where we can start to answer more complex questions:

* Does a penguin's flipper length have a relationship with its bill length?
* Do different species of penguins have different average body masses?
* How does the distribution of bill depth compare across different islands?

By visualizing two (or even more) variables simultaneously, we can uncover correlations, compare groups, and build a much richer understanding of the dataset's underlying structure. Seaborn provides powerful, high-level functions specifically designed for this purpose.

### **Numerical vs. Numerical: The Scatter Plot**

The most fundamental way to visualize the relationship between two numerical (continuous) variables is the **scatter plot**. Each point in a scatter plot represents a single observation (in our case, a single penguin), positioned according to its values on the two variables being plotted on the x and y axes.

When examining a scatter plot, we are looking for several key things:

* **Correlation:** Do the variables move together?
    * **Positive Correlation:** As the x-variable increases, the y-variable also tends to increase. The points will form a pattern that goes up and to the right.
    * **Negative Correlation:** As the x-variable increases, the y-variable tends to decrease. The points will form a pattern that goes down and to the right.
    * **No Correlation:** There is no discernible pattern between the variables. The points appear randomly scattered.
* **Strength:** How tightly clustered are the points around a potential trend? A tight cluster indicates a strong relationship, while a widely dispersed cloud of points indicates a weak one.
* **Outliers:** Are there any points that fall far away from the main cluster?

To create relational plots like scatter plots in Seaborn, we use the versatile figure-level function `sns.relplot()` (short for "relational plot"). By default, it creates a scatter plot.

Let's investigate the relationship between a penguin's `flipper_length_mm` and its `bill_length_mm`.

```python
# Create a scatter plot to explore the relationship between two numerical variables.
# `relplot` is a figure-level function for visualizing statistical relationships.
# The default `kind` is "scatter".
sns.relplot(data=penguins, x="flipper_length_mm", y="bill_length_mm")

# Add a descriptive title
plt.suptitle("Relationship between Flipper Length and Bill Length", y=1.02)

# Display the plot
plt.show()
```

#### Interpretation

* **X-axis:** Represents the flipper length in millimeters.
* **Y-axis:** Represents the bill length in millimeters.
* **Pattern:** The points generally trend upwards and to the right. This indicates a **positive correlation**: penguins with longer flippers also tend to have longer bills.
* **Structure:** The relationship doesn't appear to be a single, simple line. We can still see what looks like two or three distinct clusters of points within the overall trend. This reinforces our earlier hypothesis from the histograms. While there is an overall positive relationship, it seems to be more complex, likely influenced by another variable.

### Adding a Third Dimension with `hue`

This is where Seaborn's power truly shines. A simple scatter plot shows the relationship between two variables. But what if we suspect a third, *categorical* variable is influencing that relationship? We can encode this third variable onto our 2D plot using a visual property like color.

In Seaborn, the `hue` parameter is used for this purpose. It will automatically color the points in the scatter plot based on the category they belong to. Let's use `hue` to see if the `species` variable can explain the clusters we observed.

```python
# Enhance the scatter plot by adding a third variable, `species`, using the `hue` parameter.
# Seaborn will automatically assign a different color to each species and create a legend.
sns.relplot(
    data=penguins,
    x="flipper_length_mm",
    y="bill_length_mm",
    kind="scatter",  # Explicitly stating the kind for clarity
    hue="species"    # Color points by the 'species' column
)

# Add a title
plt.suptitle("Flipper Length vs. Bill Length by Species", y=1.02)

# Display the plot
plt.show()
```

#### Interpretation

This plot is a massive leap forward in our understanding!

* **Clarity:** The mystery of the clusters is solved. The three distinct groups of points correspond perfectly to the three penguin species: Adelie (blue), Chinstrap (orange), and Gentoo (green).
* **Species-Specific Relationships:** We can now see that the positive correlation between flipper length and bill length exists *within* each species, but the species themselves occupy very different regions of the plot.
    * **Gentoo penguins** (green) are clearly distinguished by having the longest flippers and, on average, longer bills than the Adelie.
    * **Adelie penguins** (blue) have the shortest flippers.
    * **Chinstrap penguins** (orange) fall in between on flipper length but have a wide range of bill lengths, some of which are the longest in the dataset.
* **The Power of `hue`:** By adding a single parameter, `hue='species'`, we transformed a somewhat confusing plot into a crystal-clear illustration of how three variables interact. This is a core technique in effective EDA.

### Creating Subplots with Faceting (`col` and `row`)

Another powerful technique for exploring relationships across categories is **faceting**. This involves creating multiple subplots, where each subplot shows a different subset of the data. Seaborn's figure-level functions (`relplot`, `displot`, `catplot`) make this incredibly easy with the `col` and `row` parameters.

Let's use `col="species"` to create three separate scatter plots, one for each species, arranged in columns. This can sometimes make it easier to see the specific trend within each category without the distraction of other data points.

```python
# Use the `col` parameter to create separate subplots for each species.
sns.relplot(
    data=penguins,
    x="flipper_length_mm",
    y="bill_length_mm",
    kind="scatter",
    col="species"  # Create a new column of plots for each unique value in 'species'
)

# Add a title for the entire figure
plt.suptitle("Flipper vs. Bill Length, Faceted by Species", y=1.03)

# Display the plot
plt.show()
```

This faceted plot makes the individual relationships even clearer. We can now easily compare the slope and spread of the data for Adelie, Chinstrap, and Gentoo penguins in their own dedicated plots.

### Exploring Relationships with Categorical Data

So far, we've focused on relationships between two numerical variables. But what if one of our variables is categorical? For instance, how do we compare the flipper lengths across the three different species?

For this, Seaborn provides another incredibly powerful and flexible figure-level function: `sns.catplot()` (short for "categorical plot"). Just like `relplot()`, it has a `kind` parameter that lets us create many different types of categorical visualizations.

#### Comparing Averages with Bar Plots (`kind="bar"`)

A common task is to compare the average of a numerical variable across different categories. For example, what is the *average* flipper length for each species?

We can do this by specifying both an `x` (categorical) and a `y` (numerical) variable and setting `kind="bar"`. By default, Seaborn will calculate the mean of the `y` variable for each category (this can be changed using the `estimator` parameter to show, for example, the median or sum) and display it as the height of the bar.

Notice the small black lines on top of each bar. These represent the **95% confidence interval** for the mean. This means that if we were to repeatedly sample from the population and calculate the mean flipper length for each species, 95% of the intervals constructed in this manner would contain the true population mean. It gives us a sense of the uncertainty around our estimate. Non-overlapping confidence intervals often suggest a statistically significant difference between the means of the groups.

```python
# Use catplot with kind="bar" to show the mean of a numerical variable for each category.
# The black lines are 95% confidence intervals.
sns.catplot(data=penguins, x="species", y="flipper_length_mm", kind="bar")

# Add a title
plt.suptitle("Average Flipper Length by Species", y=1.02)

# Display the plot
plt.show()
```

#### Interpretation

This plot clearly shows that, on average, Gentoo penguins have significantly longer flippers than both Adelie and Chinstrap penguins. The confidence intervals for Adelie and Chinstrap are very small and don't overlap with the Gentoo interval at all, confirming that this difference is highly unlikely to be due to random chance.

### Going Deeper: Visualizing Distributions within Categories

A bar plot is great for comparing means, but it hides a lot of information. The mean doesn't tell us anything about the range, spread, or shape of the distribution for each category. To get a better view, we can use other kinds of categorical plots.

#### The Box Plot (`kind="box"`)

A **box plot** (or box-and-whisker plot) is a classic statistical graphic that provides a five-number summary of a distribution:

* **The Median:** The line inside the box marks the median (the 50th percentile).
* **The Box:** The box itself represents the **Interquartile Range (IQR)**, spanning from the 25th percentile (Q1) to the 75th percentile (Q3).
* **The Whiskers:** The lines extending from the box (the whiskers) typically show the rest of the distribution's range, usually up to 1.5 times the IQR.
* **Outliers:** Any points that fall outside the whiskers are plotted individually.

```python
# Use kind="box" to visualize the distribution of flipper lengths for each species.
sns.catplot(data=penguins, x="species", y="flipper_length_mm", kind="box")

# Add a title
plt.suptitle("Distribution of Flipper Lengths by Species", y=1.02)

# Display the plot
plt.show()
```

Note that by swapping the `x` and `y` assignments (e.g., `x="flipper_length_mm", y="species"`), you can easily create horizontal box or violin plots, which can be useful when category names are long.

#### Interpretation

This plot is far more informative than the bar plot!

* **Gentoo:** We can see not just that Gentoo penguins have the highest *mean* flipper length, but that their entire distribution is shifted higher. The *shortest* Gentoo flipper is longer than the *median* of the other two species.
* **Adelie and Chinstrap:** The box plot shows their distributions are quite different. The Chinstrap penguins have a wider range of flipper lengths (a taller box) compared to the Adelie penguins.

#### The Violin Plot (`kind="violin"`)

A **violin plot** is a powerful hybrid that combines the best of a box plot and a KDE plot. The "violin" shape is a KDE of the data's distribution, rotated vertically. This shows the shape and density of the data at different values. Inside the violin, a miniature box plot (or sometimes just a median line) provides summary statistics. This allows us to compare both the summary and the full distribution shape at the same time.

```python
# Use kind="violin" to show both the distribution shape (KDE) and summary stats.
sns.catplot(data=penguins, x="species", y="flipper_length_mm", kind="violin")

# Add a title
plt.suptitle("Distribution of Flipper Lengths by Species", y=1.02)

# Display the plot
plt.show()
```

#### Interpretation
The violin plot confirms what we've seen before but adds another layer. The wide parts of the violin show where the data is most concentrated. For the Adelie and Chinstrap species, the violins are wider near their centers. For the Gentoo species, the violin is also wide in the middle of its range. The violin plot makes the bimodal nature of the *overall* flipper length distribution (when all species are combined) visually obvious.

## 3.5 Getting the Big Picture: Pair Plots and Heatmaps


We've done a great job drilling down into specific relationships, one or two variables at a time. But sometimes, especially during the initial phase of EDA, we want a high-level overview of *all* the relationships in our dataset at once. Seaborn provides two incredibly powerful functions for this purpose: `pairplot` and `heatmap`.

### The Ultimate Overview: `sns.pairplot()`

The `pairplot` function is a fantastic tool for visualizing the relationships between every numerical variable in your dataset. It creates a grid of axes such that each variable in your data is shared across the y-axes on a single row and the x-axes on a single column.

In the grid, you get two types of plots:

1.  **Off-diagonal:** For each pair of variables, a scatter plot is drawn. This shows you the bivariate relationship between them.
2.  **Diagonal:** Along the diagonal, where a variable would be plotted against itself, a histogram (or KDE) is drawn. This shows you the univariate distribution of that variable.

A single call to `pairplot` can generate dozens of plots, giving you a comprehensive initial summary of your data's structure. Let's create one for the penguins dataset. We can also add the `hue` parameter to see how these relationships are segmented by species.

```python
# Create a pairplot to see all pairwise relationships, colored by species.
# This can take a moment to render as it's creating many plots.
sns.pairplot(data=penguins, hue="species")
plt.suptitle("Pairwise Relationships in the Penguins Dataset", y=1.02)
plt.show()
```

### Visualizing Correlations with `sns.heatmap()`

While a `pairplot` shows us visual relationships, we often want to quantify the *strength* of these relationships. The most common way to do this for numerical variables is by calculating the **correlation matrix**. A correlation coefficient (typically Pearson's) ranges from -1 to +1:

* **+1:** Perfect positive correlation.
* **-1:** Perfect negative correlation.
* **0:** No linear correlation.

A table of these correlation values can be hard to read. A **heatmap** is the perfect visualization for this. It represents the matrix as a grid, where the color of each cell corresponds to its value. This makes it instantly easy to spot strong positive (hot) and strong negative (cold) correlations.

To create one, we first need to calculate the correlation matrix using the `.corr()` method on our DataFrame, and then pass this matrix to `sns.heatmap()`.

```python
# First, select only the numerical columns for the correlation calculation
numerical_penguins = penguins.select_dtypes(include=['float64', 'int64'])

# Calculate the correlation matrix
corr_matrix = numerical_penguins.corr()

# Create the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix of Penguin Measurements")
plt.show()
```

#### Interpretation

* `annot=True`: This argument displays the actual correlation value on each cell, making the plot more precise.
* `cmap='coolwarm'`: This color map uses warm colors (reds) for positive correlations and cool colors (blues) for negative correlations. The intensity of the color indicates the strength.
* `fmt=".2f"`: This formats the annotation numbers to two decimal places.

From the heatmap, we can instantly quantify our observations from the pairplot:

* There is a very strong positive correlation (**0.87**) between `flipper_length_mm` and `body_mass_g`.
* There is a moderate positive correlation (**0.66**) between `bill_length_mm` and `flipper_length_mm`.
* There is a moderate negative correlation (**-0.47**) between `bill_depth_mm` and `body_mass_g`.

## 4. Introduction to Altair and the Grammar of Graphics

So far, we have become proficient with Seaborn, a fantastic library for rapidly creating common statistical plots. Seaborn's functions, like `relplot()` and `catplot()`, are powerful because they are designed to produce specific, complex visualizations with a single command. This is an excellent approach for efficient Exploratory Data Analysis.

We are now going to switch gears and introduce a second, equally powerful library: **Altair**. The reason for learning Altair is not just to have another tool, but to learn a new and profoundly influential *philosophy* for creating visualizations known as the **Grammar of Graphics**.

### A New Way of Thinking: Declarative Visualization

Altair is a **declarative** statistical visualization library for Python. What does "declarative" mean?

* In a traditional, **imperative** library (like Matplotlib), you build a plot step-by-step. You might say: "Create a figure, add axes to it, plot these points as a scatter plot, set the x-axis label, set the y-axis label, add a title." You are specifying *how* to draw the plot.
* In a **declarative** library like Altair, you describe *what* you want the final plot to look like by specifying the relationships between your data and the visual elements. You don't list the steps to create it; you declare its structure.

This declarative approach is based on the "Grammar of Graphics," a formal system for describing the components of a statistical graphic. By learning this grammar, you can move beyond a fixed menu of chart types (e.g., "scatter plot," "bar chart") and gain the ability to construct a virtually limitless range of novel and insightful visualizations from a few simple building blocks.

Altair is built on top of a powerful visualization grammar called **Vega-Lite**. This means the simple Python code you write is compiled into a JSON specification that can be rendered in web browsers, which is why Altair excels at creating interactive charts, as we'll see in the next section.

### The Core Components of the Grammar

The Grammar of Graphics, as implemented in Altair, breaks every visualization down into a few core components. The three most fundamental are:

1.  **Data:** The dataset you are visualizing. In our case, this will be a pandas DataFrame. This is the raw material, the "nouns" of our visual sentence.
2.  **Mark:** The geometric shape used to represent each data record. This is the "verb"—what you are drawing. Examples include:
    * `mark_point()` for dots (used in scatter plots).
    * `mark_bar()` for rectangles (used in bar charts).
    * `mark_line()` for lines (used in line charts).
    * `mark_area()` for filled areas.
3.  **Encoding:** The set of rules that map the columns (fields) of your data to the visual properties of the marks. These are the "adjectives" or "adverbs" that describe *how* the marks should look. These visual properties are often called **channels**. Key channels include:
    * `x`: Position on the x-axis.
    * `y`: Position on the y-axis.
    * `color`: The color of the mark.
    * `size`: The size of the mark.
    * `shape`: The shape of the mark (e.g., circle, square, cross).
    * `tooltip`: Information to show when you hover over a mark.

Every chart you create in Altair will be a combination of these three elements: you take some **Data**, represent it with a **Mark**, and link the data to the mark's visual properties through an **Encoding**.

### Setting Up: A New Dataset for a New Library

To explore this new paradigm, we'll use a new dataset. The `vega_datasets` library, which is installed alongside Altair, provides a rich collection of sample data. We will use the `movies` dataset.

This dataset contains information about over 3,000 movies, including their budget, gross revenue, genre, and ratings. Let's load it and perform a small amount of preprocessing. The `Release_Date` column is stored as a string, which is not ideal for analysis. We'll write a small function to extract the year from this string and create a new `Year` column. Using `pd.to_datetime` is robust because it can parse many common date formats and gracefully handles errors, unlike simple string slicing which might break with unexpected formats.

```python
# Import the necessary libraries
import altair as alt
import pandas as pd
from vega_datasets import data as vega_data

# The vega_datasets library provides a URL to the data.
# We can load it directly into pandas.
movies_url = vega_data.movies.url
movies = pd.read_json(movies_url)

# --- Data Preprocessing ---
# The 'Release_Date' is a string like 'Jan 01 1995'. We want to extract the year.
# We'll define a function to handle this conversion safely.
def extract_year(value):
    """
    Attempts to parse a date string and return the year.
    Returns None if the format is invalid or the value is missing.
    """
    try:
        # pd.to_datetime can parse many common date formats.
        return pd.to_datetime(value).year
    except (ValueError, TypeError):
        # If a date is malformed or not a string, return None.
        return None

# Apply this function to the 'Release_Date' column to create a new 'Year' column.
movies["Year"] = movies["Release_Date"].apply(extract_year)

# Ensure the 'Title' column is a string type for consistency.
movies['Title'] = movies.Title.astype(str)

# Now, let's inspect our preprocessed DataFrame.
print("--- Movies Dataset Info ---")
movies.info()

print("\n--- First 5 Rows of Movies Dataset ---")
movies.head()
```

Our data is now ready. We have a clean `movies` DataFrame with numerical columns like `Production_Budget`, `Worldwide_Gross`, and `IMDB_Rating`, as well as categorical columns like `Major_Genre`.

### Building Our First Altair Chart: A Scatter Plot

Let's put the Grammar of Graphics into practice. Our goal is to create a scatter plot to investigate the relationship between a movie's `Production_Budget` and its `Worldwide_Gross`.

We will build this chart piece by piece, explicitly identifying the Data, Mark, and Encoding.

1.  **Data:** We start by passing our `movies` DataFrame to `alt.Chart()`. This creates the fundamental *base chart object*, which we then chain with methods like `.mark_point()` and `.encode()` to build our visualization.
2.  **Mark:** We want to represent each movie as a point, so we chain the `.mark_point()` method.
3.  **Encoding:** We define the mapping from data to visual properties with `.encode()`.
    * We map `Production_Budget` to the `x` channel.
    * We map `Worldwide_Gross` to the `y` channel.

#### Altair's Data Type Notation

You must tell Altair what *type* of data each column represents. This allows Altair to make intelligent choices about scales, axes, and legends. The type is specified with a single letter after a colon. The most common types are:

* `:Q` for **Quantitative**: Numerical data where the value and magnitude matter (e.g., budget, temperature).
* `:N` for **Nominal**: Categorical data where there is no inherent order (e.g., movie genre, country name).
* `:O` for **Ordinal**: Categorical data that has a meaningful order (e.g., "small", "medium", "large").
* `:T` for **Temporal**: Date and time values.

#### Setting Chart-Level Properties
In addition to data-driven encodings, we often want to set properties for the chart as a whole, like its title or dimensions. This is done with the `.properties()` method.

Let's see the full code.

```python
# Create a scatter plot using the Grammar of Graphics
# 1. Data: alt.Chart(movies)
# 2. Mark: .mark_point()
# 3. Encoding: .encode(...)
scatter_plot = alt.Chart(movies).mark_point().encode(
    # Map 'Production_Budget' to the x-axis, treating it as Quantitative
    alt.X('Production_Budget:Q'),

    # Map 'Worldwide_Gross' to the y-axis, treating it as Quantitative
    alt.Y('Worldwide_Gross:Q')
).properties(
    title='Production Budget vs. Worldwide Gross',
    width=500,
    height=300
).interactive()

# To display the chart in a notebook, you just call the variable.
scatter_plot
```

#### Interpretation

This chart, built from simple, declarative components, is rich with information.

* **Relationship:** There is a clear positive correlation between `Production_Budget` and `Worldwide_Gross`.
* **Density:** The vast majority of movies are clustered in the lower-left corner.
* **Outliers:** There are several points far to the right and top, representing blockbuster films.
* **Axes:** Altair has automatically created clean, readable axes with appropriate scales and labels.

### Building a Bar Chart with Aggregation

Now let's try a different visualization: a bar chart showing how many movies are in each `Major_Genre`.

1.  **Data:** `alt.Chart(movies)`
2.  **Mark:** `.mark_bar()`
3.  **Encoding:**
    * `x` channel: `Major_Genre`, a Nominal (`:N`) type.
    * `y` channel: We want the *count of movies*. Altair can perform **aggregations** directly within the encoding. Common aggregation functions include `count()`, `sum()`, `average()`, `median()`, `min()`, and `max()`. We specify `count()` as our y-variable. The result of a count is a number, so we label it as Quantitative (`:Q`).

#### Setting a Constant Value
Sometimes you want to set a visual property to a fixed value, not map it to a data column. For example, you might want all bars to be the color 'steelblue'. For this, you use `alt.value()`. This is distinct from `alt.Color('Major_Genre:N')`, which would map each genre to a *different* color.

```python
# Create a bar chart showing the count of movies per genre
bar_chart = alt.Chart(movies).mark_bar().encode(
    # Map 'Major_Genre' to the x-axis, treating it as Nominal (categorical)
    alt.X('Major_Genre:N', title='Major Genre'), # We can customize the axis title

    # For the y-axis, perform a count aggregation and treat the result as Quantitative
    alt.Y('count():Q', title='Number of Movies'),

    # We can also set a constant color for all bars
    color=alt.value('steelblue') # Use alt.value() for fixed properties
).properties(
    title='Number of Movies by Genre'
).interactive()

bar_chart
```

#### Interpretation

* **Frequency:** The height of each bar represents the number of movies in that genre.
* **Most Common Genres:** We can immediately see that "Drama", "Comedy", and "Action" are the most frequent genres.
* **Labels:** Altair has automatically labeled the x-axis with the genre names. Notice that some genres are `null`; this represents movies where the genre was not specified.

By simply changing the mark from `mark_point` to `mark_bar` and changing the encoding channels, we created a completely different but equally valid visualization. This is the power and flexibility of the Grammar of Graphics.

* Are the average ratings for most genres clustered together, or is there a wide variation?

```python
#@title Click for Solution
# Import altair just in case the cell is run independently
import altair as alt
import pandas as pd
from vega_datasets import data as vega_data

# Load and preprocess the data again for a self-contained solution
movies_url = vega_data.movies.url
movies = pd.read_json(movies_url)

def extract_year(value):
    try:
        return pd.to_datetime(value).year
    except (ValueError, TypeError):
        return None

movies["Year"] = movies["Release_Date"].apply(extract_year)
movies['Title'] = movies.Title.astype(str)


# --- Solution Code ---

# 1. Data: Start with the movies DataFrame
# 2. Mark: Use mark_bar() for a bar chart
# 3. Encoding: Map genre to x and the average rating to y
average_rating_chart = alt.Chart(movies).mark_bar().encode(
    # Map 'Major_Genre' to the x-axis as a Nominal type.
    # We can also sort the bars by the y-value for easier comparison.
    alt.X('Major_Genre:N', title='Major Genre', sort='-y'),

    # Map the average of 'IMDB_Rating' to the y-axis as a Quantitative type.
    alt.Y('average(IMDB_Rating):Q', title='Average IMDB Rating')
).properties(
    title='Average IMDB Rating by Major Genre'
)

average_rating_chart
```

**Solution Interpretation:**

* **Highest Average Rating:** The genre with the highest average IMDB rating is **"Documentary"**. This is a common finding, as documentaries are often self-selecting for audiences interested in the topic and are frequently well-regarded by critics.
* **Lowest Average Rating:** Excluding the `null` category, the genre with the lowest average IMDB rating is **"Horror"**.
* **Variation:** The average ratings for most other major genres (like Drama, Action, Comedy, Adventure) are clustered quite closely together, mostly falling between 6.0 and 6.5. This suggests that while individual movie quality varies greatly, the *average* quality across major commercial genres is fairly similar.
* **Sorting for Clarity:** The `sort='-y'` argument in the `alt.X` encoding tells Altair to sort the categories on the x-axis based on the value of the y-axis in descending order. This makes it much easier to visually compare the average ratings from highest to lowest.

## 5. Interactive Visualizations with Altair

Static visualizations are the bedrock of data analysis. However, during the *exploratory* phase, our analysis is a dynamic conversation with the data. We ask a question, create a plot, observe a pattern, and that pattern leads to a new question. This is where **interactive visualizations** become an indispensable tool.

Interactivity transforms a plot from a static statement into a dynamic exploratory interface. It allows you to:

* **Zoom** into dense regions of a plot to uncover fine-grained structure.
* **Pan** across a large dataset to explore different areas.
* **Hover** over individual data points to get precise details on demand.

One of the most compelling reasons to use Altair is that it makes creating these powerful, interactive graphics incredibly simple.

### Adding Zooming and Panning with `.interactive()`

Let's revisit our scatter plot showing the relationship between a movie's production budget and its worldwide gross, colored by genre.

```python
# Recreate our base scatter plot
base_scatter = alt.Chart(movies).mark_point().encode(
    alt.X('Production_Budget:Q'),
    alt.Y('Worldwide_Gross:Q'),
    alt.Color('Major_Genre:N')
).properties(title='Movie Budget vs. Gross (Static)')

base_scatter
```

This plot is informative, but the vast majority of points are clustered in the bottom-left corner. To investigate that dense cloud, we can add the `.interactive()` method. This single method call is a convenient shortcut that internally sets up an interval selection on the x and y axes, binding mouse controls for panning (click and drag) and zooming (scroll wheel) directly to the chart's axes.

```python
# Let's make the same chart interactive
interactive_scatter = alt.Chart(movies).mark_point().encode(
    alt.X('Production_Budget:Q'),
    alt.Y('Worldwide_Gross:Q'),
    alt.Color('Major_Genre:N')
).properties(
    title='Movie Budget vs. Gross (Interactive)'
).interactive() # This is the magic line!

interactive_scatter
```

#### Interpretation

Try interacting with the plot above!

* **Zoom:** Use your mouse's scroll wheel to zoom in on the dense cluster.
* **Pan:** Once zoomed in, click and drag the plot around. To reset, double-click.

This simple addition has fundamentally changed our ability to explore the data.

### Providing On-Demand Detail with Tooltips

Zooming is great, but what about the data points themselves? If we see an interesting outlier, how can we find out which movie it is? A **tooltip** is a small box of information that appears when you hover your mouse over a data point.

In Altair, adding a tooltip is as simple as adding another encoding channel. The `tooltip` channel takes a list of the data columns you want to display.

```python
# Add a tooltip encoding channel to our interactive scatter plot
scatter_with_tooltips = alt.Chart(movies).mark_point(filled=True).encode(
    # X, Y, and Color channels are the same as before
    alt.X('Production_Budget:Q'),
    alt.Y('Worldwide_Gross:Q'),
    alt.Color('Major_Genre:N'),

    # Add the tooltip channel, which takes a list of columns to display
    tooltip=[
        alt.Tooltip('Title:N', title='Movie Title'), # We can customize the tooltip title
        alt.Tooltip('Production_Budget:Q', format='$,.0f'), # And format the numbers
        alt.Tooltip('Worldwide_Gross:Q', format='$,.0f'),
        alt.Tooltip('IMDB_Rating:Q')
    ]
).properties(
    title='Movie Budget vs. Gross (with Tooltips)'
).interactive()

scatter_with_tooltips
```

#### Interpretation

This is now a powerful tool for exploration.

* **Hover over outliers:** Find the point at the very top of the chart. The tooltip immediately tells you it's "Avatar."
* **Explore the dense region:** Zoom into the bottom-left corner. Now, as you move your mouse around, you can see the specific details of each of those lower-budget films.

### Beyond `.interactive()`: The Power of Selections

While `.interactive()` is a fantastic shortcut, Altair's true power comes from its explicit **selection** mechanism. A selection is a data query that is generated by direct manipulation of the chart (e.g., clicking, dragging, etc.). Once defined, a selection can be used to dynamically change the plot, for example by conditionally encoding visual properties.

Let's create a simple selection that responds to a mouse click. `alt.selection_single()` creates a selection that can hold a single data point. We can then use this selection to drive a condition: we'll make the selected point bigger and all other points semi-transparent.

```python
# 1. Define a selection that triggers on click
click_selection = alt.selection_single(encodings=['x', 'y'])

# 2. Build the chart
scatter_with_selection = alt.Chart(movies).mark_point(filled=True).encode(
    alt.X('Production_Budget:Q'),
    alt.Y('Worldwide_Gross:Q'),
    tooltip=['Title:N'],

    # 3. Use the selection in a conditional encoding
    # If a point is in the selection, make its size 200, otherwise make it 50.
    size=alt.condition(click_selection, alt.value(200), alt.value(50)),

    # If a point is in the selection, make its opacity 1.0, otherwise make it 0.3.
    opacity=alt.condition(click_selection, alt.value(1.0), alt.value(0.3))

).add_selection(
    click_selection # 4. Add the selection to the chart
).properties(
    title='Click a Point to Highlight It'
)

scatter_with_selection
```

This is just a glimpse of the power of selections. They can be used to link multiple charts together, filter data on the fly, and create sophisticated, custom interactive dashboards.

### Creating an Interactive Line Chart

Let's plot the maximum daily temperature (`temp_max`) over time.

1.  **Data:** `alt.Chart(seattle_weather)`
2.  **Mark:** `mark_line()`
3.  **Encoding:**
    * `x`: The `date` column, specified as Temporal (`:T`).
    * `y`: The `temp_max` column, specified as Quantitative (`:Q`).
    * `tooltip`: To show the exact `date` and `temp_max`.
4.  **Interactivity:** `.interactive()` to allow zooming and panning.

```python
# Create an interactive line chart for time-series data
line_chart = alt.Chart(seattle_weather).mark_line(point=True).encode(
    # Map the 'date' column to the x-axis, specifying it as Temporal data
    alt.X('date:T', title='Date'),

    # Map the 'temp_max' column to the y-axis as Quantitative data
    alt.Y('temp_max:Q', title='Max Temperature (°C)'),

    # Add tooltips for specific details on hover
    tooltip=[
        alt.Tooltip('date:T', title='Date', format='%Y-%m-%d'), # Format the date
        alt.Tooltip('temp_max:Q', title='Max Temp'),
        alt.Tooltip('precipitation:Q', title='Precipitation (mm)')
    ]
).properties(
    title='Daily Max Temperature in Seattle (2012-2015)'
).interactive()

line_chart
```

Adding `point=True` to `mark_line()` ensures that individual data points are drawn on the line. This is particularly useful for interactive charts, as it makes it easier to precisely hover over a specific data point to trigger the tooltip, especially when the line itself might be thin or sparse.

Finding Local Businesses with an API

Plotting from a static file is great, but what if we want live data? We can use APIs (Application Programming Interfaces) to request data from web services. We'll use the Overpass API, a free service that lets us query the massive, open-source OpenStreetMap database. This is a powerful way to get up-to-date information about points of interest anywhere in the world without needing an API key.

We'll write a function to find all businesses of a certain type (like restaurants or bars) within a specific radius of a given point.

```python
import requests
import pandas as pd

def get_nearby_places(latitude, longitude, radius_meters, amenity_type):
    """
    Fetches nearby places from the Overpass API based on an amenity type.
    """
    overpass_url = "https://overpass-api.de/api/interpreter"

    # This is a query in the Overpass QL language.
    # It looks for nodes with a specific "amenity" tag within a certain radius.
    overpass_query = f"""
    [out:json];
    node(around:{radius_meters},{latitude},{longitude})["amenity"="{amenity_type}"];
    out;
    """

    # Use POST request, which is more robust for sending complex queries
    response = requests.post(overpass_url, data={'data': overpass_query})

    if response.status_code == 200:
        data = response.json()
        places = []
        for element in data['elements']:
            if 'tags' in element and 'name' in element['tags']:
                places.append({
                    'name': element['tags']['name'],
                    'latitude': element['lat'],
                    'longitude': element['lon']
                })
        return pd.DataFrame(places)
    else:
        print(f"Error: API request failed with status code {response.status_code}")
        print(f"Response text: {response.text}")
        return pd.DataFrame()

# Let's find restaurants and bars in Aalborg, Denmark
aalborg_lat, aalborg_lon = 57.0488, 9.9217
restaurants = get_nearby_places(aalborg_lat, aalborg_lon, 1000, "restaurant")
bars = get_nearby_places(aalborg_lat, aalborg_lon, 1000, "bar")

print(f"Found {len(restaurants)} restaurants and {len(bars)} bars.")
```

Now that we have the data in DataFrames, we can plot them on a new map. We'll use different icons and colors to distinguish between them.

```python
# Create a new map centered on Aalborg
map_aalborg = folium.Map(location=[aalborg_lat, aalborg_lon], zoom_start=14)

# Add restaurant markers (blue, with a cutlery icon)
for idx, row in restaurants.iterrows():
    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=row['name'],
        icon=folium.Icon(color='blue', icon='cutlery', prefix='fa')
    ).add_to(map_aalborg)

# Add bar markers (red, with a glass icon)
for idx, row in bars.iterrows():
    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=row['name'],
        icon=folium.Icon(color='red', icon='glass', prefix='fa')
    ).add_to(map_aalborg)

map_aalborg

```

### Seaborn: The Statistical Explorer's Toolkit

Seaborn is your go-to library for rapid, statistically-focused exploratory data analysis. Its design philosophy is centered on creating common, informative statistical graphics with as little code as possible. Think of it as a set of expertly crafted tools, each one perfectly suited for a specific analytical job.

**Key Strengths:**

* **High-Level and Concise:** Functions like `displot()`, `relplot()`, and `catplot()` are incredibly powerful, allowing you to generate complex, multi-faceted visualizations with a single, readable line of code.
* **Statistically Aware:** Seaborn's functions are built with statistical analysis in mind. They can automatically perform aggregations (like calculating the mean for a bar chart) and display measures of uncertainty (like confidence intervals) by default.
* **Deep Pandas Integration:** It is designed to work seamlessly with pandas DataFrames, making it a natural fit in the standard Python data science workflow.
* **Publication-Quality Aesthetics:** Out of the box, Seaborn produces beautiful, well-designed static plots that are suitable for reports, presentations, and academic papers.

**When to Choose Seaborn:**

* **Initial Data Exploration:** When you first receive a new dataset, Seaborn is often the fastest way to get a comprehensive overview of its distributions and the relationships between its variables.
* **Static Reporting:** If your final output is a static document like a PDF report, a slide deck, or a printed article, Seaborn's high-quality static graphics are an ideal choice.
* **Standard Statistical Plots:** When you need to create classic statistical visualizations like histograms, box plots, scatter plots with regression lines, or heatmaps, Seaborn provides a direct and efficient path.

### Altair: The Declarative Grammar for Interactive Insights

Altair introduces a different, more formal way of thinking about visualization. By embracing the Grammar of Graphics, it provides a flexible and consistent framework for building a virtually limitless variety of charts from a few simple primitives. Its true power shines when you need to create custom or interactive visualizations.

**Key Strengths:**

* **Declarative and Consistent Syntax:** The "Data, Mark, Encoding" paradigm is a powerful mental model. Once you understand it, you can construct complex charts with a clear and predictable logic, rather than memorizing dozens of different function names and parameters.
* **Effortless Interactivity:** Altair makes it trivial to add powerful interactive features like zooming, panning, and tooltips with methods like `.interactive()`. Its selection mechanism unlocks even more sophisticated interactions.
* **Web-Native Output:** Altair charts are fundamentally JSON specifications that are rendered using the Vega-Lite library in a web browser. This makes them incredibly easy to embed in websites, blogs, and web-based dashboards.
* **Compositional and Flexible:** The grammar allows you to layer, concatenate, and combine charts in sophisticated ways, giving you full control to build highly customized and novel visualizations.

**When to Choose Altair:**

* **Interactive Exploration:** When you want to build a chart that you or your audience can directly interact with—zooming into details, hovering for tooltips, or filtering by selection—Altair is the superior choice.
* **Custom Visualizations:** If you have a unique visualization design in mind that doesn't fit a standard chart type, Altair's compositional grammar gives you the building blocks to create it from scratch.
* **Web-Based Dashboards:** If your goal is to create a dashboard or an interactive report for the web, Altair's output format is perfectly suited for the task.
* **Pedagogical Clarity:** When you want your code to clearly and explicitly declare the mapping between data and visual properties, Altair's syntax is unmatched.

### A Practical Guide: Seaborn vs. Altair

| Feature               | Seaborn                                                              | Altair                                                               |
| --------------------- | -------------------------------------------------------------------- | -------------------------------------------------------------------- |
| **Philosophy** | Imperative ("function-per-plot")                                     | Declarative (Grammar of Graphics)                                    |
| **Primary Use Case** | Rapid statistical EDA, publication-quality static plots              | Interactive exploration, custom charts, web-based dashboards         |
| **Interactivity** | Limited; relies on underlying Matplotlib backends                    | Excellent; built-in and easy to implement (`.interactive()`, selections) |
| **Learning Curve** | Lower initial curve for standard plots                               | Steeper to learn the grammar, but highly consistent and powerful     |
| **Customization** | Good, but often requires dropping down to Matplotlib for fine control | Extremely flexible due to its compositional nature                   |
| **Code Example** | `sns.scatterplot(data=df, x='A', y='B')`                             | `alt.Chart(df).mark_point().encode(x='A', y='B')`                    |

**3-D plotting**

When you have a relationship where two variables determine a third (i.e., z = f(x, y)), a 2D plot just won't cut it. A 3D surface plot is perfect for visualizing this kind of data, often seen in mathematical functions, machine learning loss landscapes, or physical simulations. We'll use Matplotlib's mplot3d toolkit for this.

```python
# Import necessary libraries for 3D plotting
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

print("--- Generating a 3D Surface Plot ---")

# Create the figure and a 3D axes object
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection='3d')

# Generate the data for the plot
# Create X and Y coordinate matrices from X and Y vectors
x = np.arange(-5, 5, 0.25)
y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R) # Z is a function of X and Y

# Plot the surface
# cmap=cm.coolwarm gives it the nice red-to-blue colormap
# linewidth=0 helps in rendering a smooth surface
surf = ax.plot_surface(X, Y, Z, cmap=cm.viridis,
                       linewidth=0, antialiased=False)

# Add a color bar which maps values to colors.
fig.colorbar(surf, shrink=0.5, aspect=5)

# Add labels and a title
ax.set_title('3D Surface Plot of sin(sqrt(x^2 + y^2))', fontsize=16)
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')

plt.show()
```

Hierarchical data with sunburst graph

Sometimes, your data has a natural hierarchy, like a company's sales broken down by continent, then country, then city. A sunburst chart is perfect for this. It's like a multi-level pie chart where the inner rings represent higher levels of the hierarchy and the outer rings provide a more granular breakdown. We'll use plotly, a powerful library for interactive plotting

```python
# Install the plotly library
!pip install plotly -q

import plotly.express as px
import pandas as pd

print("\n--- Generating a Sunburst Chart ---")

# Create a sample hierarchical dataset
data = {
    "Region": ["Europe", "Europe", "Europe", "Europe", "N. America", "N. America", "N. America"],
    "Country": ["Germany", "Germany", "France", "France", "USA", "USA", "Canada"],
    "City": ["Berlin", "Munich", "Paris", "Lyon", "New York", "Chicago", "Toronto"],
    "Sales": [250, 180, 300, 120, 450, 320, 280]
}
sales_df = pd.DataFrame(data)

# Create the sunburst chart
# `path` defines the hierarchy from the center outwards
# `values` determines the size of each slice
fig = px.sunburst(
    sales_df,
    path=['Region', 'Country', 'City'],
    values='Sales',
    title='Global Sales Distribution'
)

fig.show()
```
